```python
import duckdb
import pandas as pd
import matplotlib.pyplot as plt

con = duckdb.connect("blood_data.duckdb")

query = "select YEAR(visit_date) as Year,MONTH(visit_date) as Month, count(*) as daily_donations from historical  group by YEAR(visit_date),MONTH(visit_date) order by YEAR(visit_date) ASC"

data_df = con.execute(query).df()

print(data_df)

data_df.plot(kind="line", x="Year" , y="daily_donations")
plt.show()

con.close()
```
