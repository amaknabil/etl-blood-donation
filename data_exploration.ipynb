## 1. Make data dictionary for each table

```python
import duckdb
import pandas as pd
import os
from dotenv import find_dotenv, load_dotenv


# Load environment variables
dotenv_path = find_dotenv()
load_dotenv(dotenv_path)

# --- 1. Connect to your DuckDB database ---
DUCKDB_FILE_PATH = os.getenv("DUCKDB_FILE_PATH")
TABLE_NAME = 'historical'
conn = duckdb.connect(DUCKDB_FILE_PATH)

tables = ["historical", "daily" ,"retention" ,"rate"]

for table in tables:


    # --- 2. Query your table into a pandas DataFrame ---
    df = conn.execute(f"SELECT * FROM main.{table}").fetchdf()

    # --- 3. Generate data dictionary summary ---
    summary = pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.values,
        "Unique Values": [df[col].nunique() for col in df.columns],
        "Sample Unique Values": [df[col].dropna().unique()[:5].tolist() for col in df.columns]  # preview first 5 unique values
    })

    # --- 4. Display or save the summary ---
    # print(summary)

    # save to excel
    summary.to_excel(f"data_dictionary_{table}.xlsx", index=False)
    print(f"saved {table}")


```

```python
conn.close()
```
