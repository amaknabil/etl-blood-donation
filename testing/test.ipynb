```python
import telegram
import asyncio 
from dotenv import load_dotenv,find_dotenv
import os

env_path = find_dotenv()

load_dotenv(env_path)


BOT_TOKEN = os.getenv("BOT_TOKEN")

CHANNEL_ID = os.getenv("CHANNEL_ID")

async def send_daily_message():
    """
    This function creates the bot instance and sends the messages.
    """
    bot = telegram.Bot(token=BOT_TOKEN)
    
    # --- To send a text message ---
    # This print statement is just for your console, you can change it
    print(f"Sending text to private channel {CHANNEL_ID}...")
    await bot.send_message(chat_id=CHANNEL_ID, 
                           text="Hello! This is your daily message.")
    
    print("Messages sent successfully!")

# This is the standard way to run an "async" function in Python
if __name__ == "__main__":
    asyncio.run(send_daily_message())
```
