```python
# venv\Scripts\activate

import duckdb
import pandas as pd
import os
from dotenv import find_dotenv, load_dotenv

# Load environment variables
dotenv_path = find_dotenv()
load_dotenv(dotenv_path)

# 1. Define the variables
DUCKDB_FILE_PATH = os.getenv("DUCKDB_FILE_PATH")
TABLE_NAME = 'historical'

# 2. Connect to the DuckDB file
try:
    con = duckdb.connect(database=DUCKDB_FILE_PATH, read_only=True)

    # 3. Define the SQL query
    # The full path to your table is main.historical
    query = f"SELECT * FROM main.{TABLE_NAME}" 

    # 4. Execute the query and fetch the result into a Pandas DataFrame
    df_historical = con.execute(query).fetchdf()
    
    # 5. Close the connection
    con.close()

    print(f"Successfully loaded data from the '{TABLE_NAME}' table.")
    print(df_historical.head())

except Exception as e:
    print(f"An error occurred: {e}")
    print(f"Please ensure the file '{DUCKDB_FILE_PATH}' exists and is a valid DuckDB database.")
```
